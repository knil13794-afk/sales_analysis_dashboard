```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("Cleaned_Sales.csv")
df
```

```python
sales_region=df.groupby("region")["sales_amount"].sum()

sales_region.plot(kind='bar')

plt.title("Sales by Region")
plt.xlabel("Region")
plt.ylabel("Sales_Amount")

plt.show()
```

```python
sales_region=df.groupby("order_date")["sales_amount"].sum()

sales_region.plot(kind='line')

plt.title("Sales Trend")
plt.xlabel("Order_date")
plt.ylabel("Sales_Amount")

plt.show()
```

```python
df["sales_amount"].plot(kind='line')

plt.title("Sales Trend")
plt.xlabel("Index")
plt.ylabel("Sales_Amount")

plt.show()
```

```python
category_sales=df.groupby("category")["sales_amount"].sum()
category_sales.plot(kind='pie',autopct='%1.1f%%')

plt.title("Sales Distribution")
plt.ylabel("")

plt.show()
```

```python
plt.hist(df["sales_amount"])

plt.title("Sales Distribution")
plt.xlabel("Sales_Amount")
plt.ylabel("Frequency")

plt.show()
```

```python
df['bonus']=df['sales_amount']*0.10
print(df)
```

```python
df.to_csv("Cleaned_Sales.csv")
df
```

```python
plt.scatter(df["sales_amount"],df["bonus"])

plt.title("Sales vs Bonus")
plt.xlabel("Sales_Amount")
plt.ylabel("Bonus")

plt.show()
```

```python
plt.savefig("Sales_Chart.png")
```
